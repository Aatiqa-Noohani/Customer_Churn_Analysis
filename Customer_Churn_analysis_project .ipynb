```python
plan_churn = pd.crosstab(
    df["Subscription_Plan"],
    df["Churn_Status"]
)

print(plan_churn)
```

```python
plan_churn_rate = (
    df.groupby("Subscription_Plan")["Churn_Flag"]
    .mean()
    .mul(100)
    .round(2)
    .sort_values(ascending=False)
)

print(plan_churn_rate)
```

```python
import matplotlib.pyplot as plt

plan_churn_rate.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Churn Rate by Subscription Plan")
plt.xlabel("Subscription Plan")
plt.ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# **Business Insight**

Churn patterns differ across subscription plans and Auto Renewal status. Comparing these two dimensions helps identify specific customer segments with relatively higher or lower churn rates. For example, if a particular plan has a higher churn rate among customers without Auto Renewal, that segment can be monitored for renewal friction, payment issues, or engagement problems.
